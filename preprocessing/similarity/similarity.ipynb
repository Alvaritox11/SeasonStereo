# Similarity

Inspect one pair, its disparity, and the right image warped into the left frame.

```python
from pathlib import Path
import matplotlib.pyplot as plt
from metrics_common import load_image, load_disp, prepare_warped_pair

left_path = Path("data/synchronic_only/L/example.iio")
right_path = Path("data/synchronic_only/R/example.iio")
disp_path = Path("outputs/similarity/disparities/monster++/example.iio")

left = load_image(left_path)
right = load_image(right_path)
disp = load_disp(disp_path)
prepared = prepare_warped_pair(left, right, disp, require_positive_warp=True)

fig, axes = plt.subplots(1, 4, figsize=(18, 4))
axes[0].imshow(left.astype("uint8"))
axes[0].set_title("left")
axes[1].imshow(right.astype("uint8"))
axes[1].set_title("right")
axes[2].imshow(prepared["left_hat"].clip(0, 255).astype("uint8"))
axes[2].set_title("right warped")
axes[3].imshow(prepared["final_mask"], cmap="gray")
axes[3].set_title("valid mask")
for ax in axes:
    ax.axis("off")
plt.tight_layout()
```
