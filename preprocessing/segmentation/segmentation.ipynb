# Segmentation

Small visual check for one image, one generated mask, and one probability map.

```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image

image_path = Path("data/Train-Track3-cropped/Track3-RGB-1/JAX_001/JAX_001_001_RGB.tif")
mask_path = Path("data/water_segmentation/masks/JAX_001_001_RGB_mask.png")
prob_path = Path("data/water_segmentation/probs/JAX_001_001_RGB_water_prob.png")

image = np.array(Image.open(image_path).convert("RGB"))
mask = np.array(Image.open(mask_path))
prob = np.array(Image.open(prob_path)) if prob_path.exists() else None

cols = 3 if prob is not None else 2
fig, axes = plt.subplots(1, cols, figsize=(5 * cols, 5))
axes[0].imshow(image)
axes[0].set_title("image")
axes[1].imshow(mask, cmap="gray")
axes[1].set_title("mask")
if prob is not None:
    axes[2].imshow(prob, cmap="magma")
    axes[2].set_title("probability")
for ax in axes:
    ax.axis("off")
plt.tight_layout()
```
